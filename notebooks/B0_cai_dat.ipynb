## 4. Tải trước mô hình (khoảng 3 GB, vài phút khi có mạng)

Tải trước để lúc chạy B2–B4 không phải chờ. Bỏ bớt tên nào không cần.

```python
from transformers import AutoModel, AutoTokenizer, AutoImageProcessor
for ten in ["clip", "dinov2", "siglip"]:
    AutoModel.from_pretrained(L.IMG_MODELS[ten]); AutoImageProcessor.from_pretrained(L.IMG_MODELS[ten]); print("ok", ten)
for ten in ["e5", "e5_b", "bkai"]:
    AutoModel.from_pretrained(L.TXT_MODELS[ten]); AutoTokenizer.from_pretrained(L.TXT_MODELS[ten]); print("ok", ten)
```
